# Scheduler benchmark figures

```python
import helpers as h

h.setup_style()
df = h.load()
# the proof scenarios end at a known goal cost, the optimization ones just minimize it
PROOF = ("caviar",)
OPT = ("integ", "trig", "szalinski")
proof = df.filter(h.pl.col("case").is_in(PROOF))
opt = df.filter(h.pl.col("case").is_in(OPT))
N = len(OPT)
```

```python
from typing import cast

KS = (1, 2, 5, 20, 100, 200, 300, 400, 500)
LHS_RHS = cast(
    tuple[h.Scheduler, ...],
    ("backoff", *(f"detour-{side}-{k}" for side in ("lhs", "rhs") for k in KS)),
)
assert set(LHS_RHS) <= set(h.SCHEDULERS), set(LHS_RHS) - set(h.SCHEDULERS)
h.check_runs(df, schedulers=LHS_RHS)
```

### Ranking

```python
h.save(h.ranking(opt, LHS_RHS), "opt-ranking", (4 * N, 4.5))
```

### Quality over resources

```python
h.save(h.quality(opt, LHS_RHS, log_x=True, log_y=True), "opt-quality", (4 * N, 6))
```

```python
h.save(h.dev(opt, LHS_RHS), "opt-dev", (4 * N, 8.5))
```
