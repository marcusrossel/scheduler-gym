## Best k per case study

### Suggested ks
The ks of each case study, ranked for lhs and rhs by geomean final cost / best final cost among `LHS_RHS` (for caviar, by most solved runs first).

```python
with h.pl.Config(tbl_rows=-1):
    display(h.suggest_best_k(df, LHS_RHS))
```

```python
# per case study, the k shown as lhs-best / rhs-best; see the suggestion above
BEST_K = {
    "caviar": {"lhs": 300, "rhs": 300},
    "integ": {"lhs": 300, "rhs": 400},
    "szalinski": {"lhs": 200, "rhs": 300},
    "trig": {"lhs": 200, "rhs": 100},
}
BEST: tuple[h.Shown, ...] = ("backoff", "lhs-best", "rhs-best")
best = h.with_best_k(df, BEST_K)
best_caviar = best.filter(h.pl.col("case").str.starts_with("caviar:"))["case"][0]
best_no_caviar = best.filter(h.pl.col("case") != best_caviar)
h.check_runs(best, BEST)
```

### Ranking

```python
h.save(h.ranking(best_no_caviar, BEST), "best-k-ranking", (4 * N, 3))
```

```python
h.save(h.sums(best_no_caviar, BEST), "best-k-sums", (4 * N, 6.5))
```

### Quality over resources

```python
h.save(
    h.quality(best_no_caviar, BEST, log_x=True, log_y=True),
    "best-k-quality",
    (4 * N, 6),
)
```

```python
h.save(h.dev(best_no_caviar, BEST), "best-k-dev", (4 * N, 8.5))
```

```python
h.save(h.profile(best_no_caviar, BEST), "best-k-profile", (4 * N, 6))
```

### Caviar

```python
h.save(h.solved(best, BEST, case=best_caviar), "best-k-caviar-solved", (6, 2))
```

```python
h.save(
    h.solved_profile(best, BEST, case=best_caviar, y_min=0.8),
    "best-k-caviar-profile",
    (10, 3.5),
)
```
